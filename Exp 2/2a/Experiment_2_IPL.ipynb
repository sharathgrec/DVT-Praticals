```python
import pandas as pd
import sqlite3

ipl = pd.read_csv("IPL_ball_by_ball_updated.csv")
print("IPL CSV file imported successfully.")
print("Rows:", len(ipl))
print("Columns:", len(ipl.columns))
```

```python
display(ipl.head())
```

```python
ipl.info()
```

```python
display(ipl.describe(include="all").T)
```

```python
ipl_sample = ipl.head(20)
ipl_sample.to_excel("ipl_sample.xlsx", index=False)

ipl_excel = pd.read_excel("ipl_sample.xlsx")
print("Excel file imported successfully.")
display(ipl_excel.head())
```

```python
conn = sqlite3.connect("ipl_data.db")
ipl_sample.to_sql("ipl_deliveries", conn, if_exists="replace", index=False)

query = "SELECT * FROM ipl_deliveries LIMIT 5"
ipl_sql = pd.read_sql(query, conn)

print("SQL data imported successfully.")
display(ipl_sql)
conn.close()
```

```python
ipl_sample.to_csv("ipl_export.csv", index=False)
print("IPL CSV file exported successfully.")
print("Exported rows:", len(ipl_sample))
```

```python
ipl_sample.to_excel("ipl_export.xlsx", index=False)
print("IPL Excel file exported successfully.")
print("Exported rows:", len(ipl_sample))
```

```python
ipl_sample.to_json("ipl_export.json", orient="records", indent=2)
print("IPL JSON file exported successfully.")
print("Exported records:", len(ipl_sample))
```

```python
conn = sqlite3.connect("ipl_export.db")
ipl_sample.to_sql("ipl_data", conn, if_exists="replace", index=False)

count = pd.read_sql("SELECT COUNT(*) AS row_count FROM ipl_data", conn)
print("IPL data exported to SQLite successfully.")
display(count)
conn.close()
```
