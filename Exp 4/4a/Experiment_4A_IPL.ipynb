```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

students = pd.DataFrame({
    "Student":["Arun","Bala","Divya","Priya","Ravi","Karthik"],
    "Department":["AI&DS","CSE","AI&DS","ECE","CSE","ECE"],
    "Marks":[88,76,92,81,69,95]
})

print("Dataset loaded successfully")
print("Shape:", students.shape)
students.head()
```

```python
print("Last records:")
students.tail()
```

```python
print("Data types:")
print(students.dtypes)
print("\nStatistical summary:")
students.describe(include="all")
```

```python
filtered = students[(students["Marks"] >= 80) & (students["Department"] == "AI&DS")]
print("AI&DS students scoring 80 or above:")
filtered
```

```python
print("Students sorted by marks:")
students.sort_values("Marks", ascending=False)
```

```python
print("Department-wise statistical analysis:")
students.groupby("Department")["Marks"].agg(["mean","median","min","max","std"]).round(2)
```

```python
# Correlation of the numerical attribute
print("Correlation matrix:")
students[["Marks"]].corr()
```

```python
plt.figure(figsize=(8,5))
plt.bar(students["Student"], students["Marks"])
plt.title("Student-wise Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))
students.groupby("Department")["Marks"].sum().plot(kind="bar")
plt.title("Department-wise Total Marks")
plt.xlabel("Department")
plt.ylabel("Total Marks")
plt.tight_layout()
plt.show()
```
