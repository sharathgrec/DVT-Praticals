```python
import pandas as pd

df = pd.read_csv("IPL_ball_by_ball_updated.csv")

print("First 5 records of IPL dataset:")
display(df.head())

print("Dataset Shape:", df.shape)
print("Number of Columns:", len(df.columns))
```

```python
import matplotlib.pyplot as plt

season_runs = df.groupby("season")["runs_off_bat"].sum()

plt.figure(figsize=(10, 5))
plt.plot(season_runs.index, season_runs.values, marker="o")
plt.title("IPL Total Runs by Season")
plt.xlabel("Season")
plt.ylabel("Total Runs Off Bat")
plt.xticks(season_runs.index, rotation=45)
plt.tight_layout()
plt.show()
```
